```python
import getpass
from google import genai
from google.colab import userdata

# 1. Read the API key from Colab Secrets, or prompt if not found
try:
    api_key = userdata.get("GEMINI_API_KEY")
except Exception:
    print("GEMINI_API_KEY not found in Colab Secrets.")
    print("Please click the key icon (🔑) on the left panel to add it for future use.")
    api_key = getpass.getpass("Please enter your Gemini API Key now: ")

if not api_key:
    raise ValueError("API Key is required to run this notebook.")

# 2. Create the Google GenAI API client
client = genai.Client(api_key=api_key)

# 3. Write a prompt for the model
prompt = """Explain Artificial Intelligence in simple words.
Give five real-world examples suitable for a
first-year engineering student."""

# Using the lightweight 'gemini-2.5-flash-lite' model as requested
try:
    response = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=prompt
    )
    # 4. Print the generated response
    print("\n========== LLM RESPONSE ========== \n")
    print(response.text)
    print("\n==================================")
except Exception as e:
    print(f"Generation failed: {e}")
```
